```python
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)
z = x * y + 2
```

```python
def demonstrate_computation_graph():
    # 创建叶子节点
    a = torch.tensor(2.0, requires_grad=True)
    b = torch.tensor(3.0, requires_grad=True)
    c = torch.tensor(4.0, requires_grad=True)

    print("叶子节点：")
    print(f"a: {a}, grad_fn: {a.grad_fn}") # None -用户创建
    print(f"b: {b}, grad_fn: {b.grad_fn}") # None
    print(f"c: {c}, grad_fn: {c.grad_fn}") # None

    # 执行运算-构建计算图
    d = a * b
    e = d + c
    f = e ** 2

    print("\n中间节点：")
    print(f"d = a * b: {d}, grad_fn: {d.grad_fn}") # MulBackward
    print(f"e = d + c: {e}, grad_fn: {e.grad_fn}") # AddBackward
    print(f"f = e ^ 2: {f}, grad_fn: {f.grad_fn}") # PowBackward

    return f, a, b, c
f, a, b, c = demonstrate_computation_graph()
```

```python
# 计算图的可视化demo
def visualize_graph():
    x = torch.tensor(1.0, requires_grad=True)
    y = torch.tensor(2.0, requires_grad=True)

    # 构建复杂计算图
    u = x * y
    v = torch.sin(u)
    w = v + x
    z = w ** 2

    print("计算图结构：")
    current = z
    while current.grad_fn is not None:
        print(f"{current} <- {current.grad_fn}")
        current = current.grad_fn.next_functions[0][0] if current.grad_fn.next_functions else None

    return z

z = visualize_graph()
```

```python
def backward_propagation_example():
    # 创建计算图
    x = torch.tensor(2.0, requires_grad=True)
    y = torch.tensor(3.0, requires_grad=True)

    z = x ** 2 + y ** 3
    print(f"z={z}")

    # 反向传播
    z.backward()

    print(f"∂z/∂x = {x.grad}")
    print(f"∂z/∂y = {y.grad}")
backward_propagation_example()
```
